# ML PIPELINE

### Importing necessary libraries

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score, precision_score, recall_score,f1_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

### Splitting the dataset into Input and Targets

```python
X = df.drop(columns=["customerID","Churn"])
y = df["Churn"]
```

## Train test split the data

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
```

### Identifying numerical and categorical features

```python
numeric_cols = X_train.select_dtypes(include=["int64","float64"]).columns.tolist()
categorical_cols = X_train.select_dtypes(include=["object", "category"]).columns.tolist()

print("Numerical columns:")
print(numeric_cols)

print("\nCategorical columns:")
print(categorical_cols)
```

### Building Preprocessing Pipeline

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_cols),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols)
    ]
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Training data shape:", X_train_processed.shape)
print("Testing data shape:", X_test_processed.shape)
```

```python
print("Original feature count:", X_train.shape[1])
print("Processed feature count:", X_train_processed.shape[1])

feature_names = preprocessor.get_feature_names_out()
print(feature_names)
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train_processed, y_train)
```

### Model evaluation

```python
print("Training Efficiency:", model.score(X_train_processed, y_train))
print("Testing Efficiency:", model.score(X_test_processed, y_test))
```

### Another method of looking at model score

```python
y_train_pred = model.predict(X_train_processed)

# Predict on the testing data
y_test_pred = model.predict(X_test_processed)

# Calculate training accuracy
train_accuracy = accuracy_score(y_train, y_train_pred)

# Calculate testing accuracy
test_accuracy = accuracy_score(y_test, y_test_pred)

print(f"Training Accuracy: {train_accuracy:.4f}")
print(f"Testing Accuracy: {test_accuracy:.4f}")
```

### Calculating the Generic accuracy

```python

print("Accuracy:", accuracy_score(y_test, y_test_pred))
print("Precision:", precision_score(y_test, y_test_pred, pos_label="Yes"))
print("Recall:", recall_score(y_test, y_test_pred, pos_label="Yes"))
print("F1 score:", f1_score(y_test, y_test_pred, pos_label="Yes"))
```

```python
confusion_matrix(y_test, y_test_pred)
```

# Train the balanced Logistic Regression model

```python

model_balanced = LogisticRegression(
    max_iter=1000,
    class_weight="balanced"
)

model_balanced.fit(X_train_processed, y_train)
```

```python
y_test_pred_balanced = model_balanced.predict(X_test_processed)

print("Accuracy:", accuracy_score(y_test, y_test_pred_balanced))
print("Precision:", precision_score(y_test, y_test_pred_balanced, pos_label="Yes"))
print("Recall:", recall_score(y_test, y_test_pred_balanced, pos_label="Yes"))
print("F1 score:", f1_score(y_test, y_test_pred_balanced, pos_label="Yes"))

# Confusion matrix
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_test_pred_balanced, labels=["No", "Yes"]))
```

# Creating the Final Pipeline

```python
final_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        max_iter=1000,
        class_weight="balanced"
    ))
])

# Train the complete pipeline
final_pipeline.fit(X_train, y_train)
print("Final pipeline trained successfully!")
```

```python
y_pred_final = final_pipeline.predict(X_test) # The X_test will undergo all same procedure from which the X_train went

print("Accuracy:", accuracy_score(y_test, y_pred_final))
print("Precision:", precision_score(
    y_test, y_pred_final, pos_label="Yes"
))
print("Recall:", recall_score(
    y_test, y_pred_final, pos_label="Yes"
))
print("F1-score:", f1_score(
    y_test, y_pred_final, pos_label="Yes"
))

print("\nConfusion Matrix:")
print(confusion_matrix(
    y_test, y_pred_final, labels=["No", "Yes"]
))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_final))
```

```python

import joblib

# Save the complete trained pipeline
joblib.dump(final_pipeline, "churn_pipeline.joblib")

print("Pipeline saved successfully!")
```
